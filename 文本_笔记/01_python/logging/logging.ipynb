````python
import logging

logging.basicConfig(filename='example.log',  # 记录到文件
                    encoding='utf-8',  # 日志编码格式
                    level=logging.DEBUG,  # 日志记录级别
                    filemode='w',  # 每次日志记录都截断
                    format='%(levelname)s:%(message)s:%(levelname)s:%(levelno)s',  # 日志消息格式
                    datefmt='%m/%d/%Y %I:%M:%S %p'  # 对日志消息中时间格式的描述
                    )

logging.info("asd")
# 类型判断
isinstance("", int)
````

对于 `logging.basicConfig`的 `format` 各个属性介绍
---
%(asctime)s     易读的时间

%(created)s     创建的时间

%(filename)s    文件名

%(levelname)s   记录级别

%(levelno)s     记录级别(数字)

%(lineno)d      行号

%(module)s      当前模块名称

%(name)s        日志记录器名称

%(pathname)s    发出日志记录调用的源文件的完整路径名

%(process)d     进程ID

%(processName)s 进程名

%(thread)d      线程ID

%(threadName)s  线程名

%(relativeCreated)d  以毫秒数表示的 LogRecord 被创建的时间

  ID:
    name:
  ID_1:
    name:
handlers:
  name:
    class: "logging.StreamHandler"  # 处理器类的完整限定名称(强制)
    level:
    formatter:  # 处理器对应的 格式化器的 ID
    filters:  # 处理器对应的   过滤器的 ID 的列表
    stream: "ext://sys.stdout"  # 输出
root:
loggers:
  name:
    level:
    propagate:  # 传播设置.  0: 不传播给父级; 1: 传播给父级
    filters:
    handlers:  # list[]
incremental:  # 配置是否要被解读为在现有配置上新增
disable_existing_loggers:  # 是否要禁用任何现有的非根日志记录器
```
